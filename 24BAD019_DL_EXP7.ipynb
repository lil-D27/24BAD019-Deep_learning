Part A: Dataset Preparation and Text Preprocessing

```python
!pip -q install nltk gensim

import nltk
import string
from nltk.corpus import brown, stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

nltk.download('brown', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)

original_sentences = brown.sents(categories='news')[:4000]
print('Total sentences:', len(original_sentences))
print('Original sentence:', ' '.join(original_sentences[0]))
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
clean_sentences = []

for sentence in original_sentences:
    text = ' '.join(sentence).lower()             # Lowercase
    words = word_tokenize(text, preserve_line=True) # Tokenize
    words = [word for word in words if word.isalpha()] # Remove punctuation/numbers
    words = [word for word in words if word not in stop_words] # Remove stop words
    words = [lemmatizer.lemmatize(word) for word in words] # Lemmatize
    if words:
        clean_sentences.append(words)

print('Original:', ' '.join(original_sentences[0]))
print('Processed:', ' '.join(clean_sentences[0]))
print('Number of processed sentences:', len(clean_sentences))
print("Dhanya G - 24BAD019")
```

Part B: Vocabulary and Word Frequency

```python
from nltk import FreqDist

all_words = [word for sentence in clean_sentences for word in sentence]
frequencies = FreqDist(all_words)

print('Vocabulary size:', len(frequencies))
print('\nTop 10 words:')
for word, count in frequencies.most_common(10):
    print(word, ':', count)

print('\nWord2Vec training sentence:', clean_sentences[0])
print("Dhanya G - 24BAD019")
```

Part C: Generate Word Embeddings using Word2Vec

```python
from gensim.models import Word2Vec

model = Word2Vec(
    sentences=clean_sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=1,              # Skip-gram
    workers=2,
    epochs=10,
    seed=42
)

word = 'government'
print('Word:', word)
print('Embedding vector:', model.wv[word])
print('Vector dimensions:', len(model.wv[word]))

print('\nFive most similar words:')
for similar_word, score in model.wv.most_similar(word, topn=5):
    print(similar_word, ':', round(score, 3))
print("Dhanya G - 24BAD019")
```

Part D: Word Similarity Analysis and Experiment Management

```python
# Use pairs whose words are present in the model vocabulary
pairs = [
    ('government', 'state'),
    ('president', 'government'),
    ('school', 'student')
]

results = []
for word1, word2 in pairs:
    if word1 in model.wv and word2 in model.wv:
        score = model.wv.similarity(word1, word2)
        results.append((word1, word2, score))
        print(word1, '-', word2, ':', round(float(score), 3))
    else:
        print('Not in vocabulary:', word1, word2)

if results:
    best = max(results, key=lambda item: item[2])
    print('\nMost similar pair:', best[0], 'and', best[1])
    print('Similarity score:', round(float(best[2]), 3))
print("Dhanya G - 24BAD019")
```
